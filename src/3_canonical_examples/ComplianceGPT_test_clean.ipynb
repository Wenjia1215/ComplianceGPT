# ComplianceGPT Canonical Examples

Minimal notebook for running the 9 fixed canonical prompts through ComplianceGPT and saving **full answers** to Drive.

Outputs are written to:

`/content/drive/MyDrive/ComplianceGPT/src/3_canonical_examples/results`


```python
from google.colab import drive
drive.mount('/content/drive')

import os
import sys
import json
import pandas as pd
from datetime import datetime
from IPython.display import display

PROJECT_ROOT = "/content/drive/MyDrive/ComplianceGPT"
SRC_DIR = os.path.join(PROJECT_ROOT, "src")
RESULTS_DIR = "/content/drive/MyDrive/ComplianceGPT/src/3_canonical_examples/results"

os.makedirs(RESULTS_DIR, exist_ok=True)

if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

print("[OK] Paths configured.")
print("PROJECT_ROOT:", PROJECT_ROOT)
print("RESULTS_DIR:", RESULTS_DIR)
```

```python
from compliancegpt.pipeline.pipeline import ComplianceGPTPipeline
from compliancegpt.retriever.retriever_s7 import RetrievalConfig

FRAMEWORK_VERSION = "rev5"
USE_ORG_PROFILE = False
MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"
TOP_K = 12

REV5_CATALOG_PATH = os.path.join(PROJECT_ROOT, "data/ccs/nist800-53/NIST_SP-800-53_rev5_catalog.jsonl")
REV4_CATALOG_PATH = os.path.join(PROJECT_ROOT, "data/ccs/nist800-53/NIST_SP-800-53_rev4_catalog.jsonl")
ORG_PROFILE_REV5 = os.path.join(PROJECT_ROOT, "data/ODP/rev5/org_profile_example_rev5.yaml")
ORG_PROFILE_REV4 = os.path.join(PROJECT_ROOT, "data/ODP/rev4/org_profile_example_rev4.yaml")

fw = FRAMEWORK_VERSION.strip().lower()
if fw not in {"rev4", "rev5"}:
    raise ValueError(f"FRAMEWORK_VERSION must be 'rev4' or 'rev5' (got {FRAMEWORK_VERSION!r})")

ccs_path = REV5_CATALOG_PATH if fw == "rev5" else REV4_CATALOG_PATH
org_profile = (ORG_PROFILE_REV5 if fw == "rev5" else ORG_PROFILE_REV4) if USE_ORG_PROFILE else None

retr_cfg = RetrievalConfig()

pipe = ComplianceGPTPipeline(
    framework_version=fw,
    model_id=MODEL_ID,
    ccs_path=ccs_path,
    use_qur=True,
    doc_filter_mode="prefer_smt_keep_params",
    org_profile_path=org_profile,
    retriever_config=retr_cfg,
    verify_strict_extras=False,
    verify_strict_verbatim=True,
    verify_strict_version=False,
)

print("[OK] ComplianceGPT pipeline initialized.")
print("CCS:", ccs_path)
```

```python
CANONICAL_EXAMPLES = {
    "odp_password_length": [
        "What is the required password length in NIST SP 800-53?",
        "Under NIST SP 800-53, how long do passwords need to be?",
        "Based on NIST SP 800-53, what password length is required?",
    ],
    "language_mismatch_mfa_admins": [
        "Do we need MFA for admins under NIST SP 800-53?",
        "Under NIST SP 800-53, do administrators need multi-factor authentication?",
        "Do privileged users have to use MFA in NIST SP 800-53?",
    ],
    "posthoc_citation_privileged_mfa": [
        "Do privileged users need multi-factor authentication? Please include the control citation.",
        "Which NIST SP 800-53 control says privileged users need MFA? Please answer with the citation.",
        "Please answer using NIST SP 800-53 and include the control citation: Do admins need MFA?",
    ],
}

print("[OK] Canonical examples loaded.")
for k, v in CANONICAL_EXAMPLES.items():
    print(f"- {k}: {len(v)} prompts")
```

```python
rows = []

for example_name, prompts in CANONICAL_EXAMPLES.items():
    print("\n" + "=" * 100)
    print(f"EXAMPLE: {example_name}")
    print("=" * 100)
    for variant_idx, prompt in enumerate(prompts, start=1):
        print(f"\n[{variant_idx}/{len(prompts)}] {prompt}")
        out = pipe.answer(
            prompt,
            top_k=TOP_K,
            use_generator=True,
            run_verify=False,
        )

        dbg = out.get("debug") or {}
        rewrites = dbg.get("rewrites") or []
        evidence_spans = out.get("evidence_spans") or []
        answer_text = str(out.get("answer_text", "") or "")

        print("status:", out.get("status"))
        print("primary_citation:", out.get("primary_citation"))
        print("odp_required_list:", out.get("odp_required_list") or [])
        print("evidence_spans_len:", len(evidence_spans))
        print("answer:")
        print(answer_text)

        rows.append({
            "framework_version": FRAMEWORK_VERSION,
            "use_org_profile": USE_ORG_PROFILE,
            "model_id": MODEL_ID,
            "example_name": example_name,
            "variant_idx": variant_idx,
            "prompt": prompt,
            "status": out.get("status"),
            "primary_citation": out.get("primary_citation"),
            "odp_required_list": json.dumps(out.get("odp_required_list") or [], ensure_ascii=False),
            "evidence_spans_len": len(evidence_spans),
            "answer_text": answer_text,
            "evidence_spans": json.dumps(evidence_spans, ensure_ascii=False),
            "rewrites_count": len(rewrites),
            "rewrites": json.dumps(rewrites, ensure_ascii=False),
        })

df_results = pd.DataFrame(rows)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
csv_path = os.path.join(RESULTS_DIR, f"compliancegpt_canonical_examples_{FRAMEWORK_VERSION}_{timestamp}.csv")
json_path = os.path.join(RESULTS_DIR, f"compliancegpt_canonical_examples_{FRAMEWORK_VERSION}_{timestamp}.json")
latest_csv_path = os.path.join(RESULTS_DIR, f"compliancegpt_canonical_examples_{FRAMEWORK_VERSION}_latest.csv")
latest_json_path = os.path.join(RESULTS_DIR, f"compliancegpt_canonical_examples_{FRAMEWORK_VERSION}_latest.json")

df_results.to_csv(csv_path, index=False)
df_results.to_csv(latest_csv_path, index=False)

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(rows, f, ensure_ascii=False, indent=2)
with open(latest_json_path, "w", encoding="utf-8") as f:
    json.dump(rows, f, ensure_ascii=False, indent=2)

print("\n[OK] Batch run complete. Rows:", len(df_results))
print("[OK] Saved CSV:", csv_path)
print("[OK] Saved JSON:", json_path)

display(df_results[[
    "example_name", "variant_idx", "prompt", "status",
    "primary_citation", "odp_required_list", "answer_text"
]])
```
